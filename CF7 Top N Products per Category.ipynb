```python
import pandas as pd
```

```python
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_excel('/content/drive/My Drive/Colab Notebooks/Superstore.xlsx')
df.head()
```

```python
df2 = df.groupby(['Category','Product_Name'])['Sales'].sum().reset_index(name='Total_Sales')
df2.head()
```

```python
df2['Rank'] = (df2.groupby('Category')['Total_Sales'].rank(method = 'min', ascending = False))
df2
```

```python
top3 = df2[df2['Rank'] <= 3].sort_values(['Category','Rank']).reset_index(drop=True)
top3
```
