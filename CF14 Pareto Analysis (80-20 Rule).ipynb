```python
import pandas as pd
```

```python
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_excel('/content/drive/My Drive/Colab Notebooks/Superstore.xlsx')
df.head()
```

```python
CustomerSales = df.groupby(['Customer_Name','Customer_ID'])['Sales'].sum().reset_index(name='TotalSales')
CustomerSales.head()
```

```python
CustomerSales = CustomerSales.sort_values(by = 'TotalSales', ascending = False).reset_index(drop=True)
CustomerSales
```

```python
grand_Total= CustomerSales['TotalSales'].sum()
grand_Total
```

```python
CustomerSales['RunningSales']= CustomerSales['TotalSales'].cumsum().round(2)
CustomerSales
```

```python
CustomerSales['CumulativePctSales'] = CustomerSales['RunningSales']/grand_Total
CustomerSales
```

```python
CustomerSales['CustomerRank'] = CustomerSales.index + 1
CustomerSales
```

```python
TotalCustomers = len(CustomerSales)
TotalCustomers
```

```python
CustomerSales['CumulativePctCustomers']= (CustomerSales['CustomerRank']/TotalCustomers *100).round(2)
CustomerSales
```

```python
CustomerSales['ParetoGroup'] = CustomerSales['CumulativePctCustomers'].apply(lambda x: 'Top 80%' if x <= 80 else 'Botton 20%')
CustomerSales
```
