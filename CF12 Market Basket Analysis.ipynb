```python
import pandas as pd
from itertools import combinations
from collections import Counter
```

```python
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_excel('/content/drive/My Drive/Colab Notebooks/Superstore.xlsx')
df.head()
```

```python
order_products = df.groupby('Order_ID')['Product_Name'].apply(lambda x: sorted(set(x))).reset_index()
order_products
```

```python
pair_counts = Counter()
for products in order_products:
  if len(products) > 1:
    for pair in combinations(products,2):
      pair_counts[pair] += 1
```

```python
basket = pd.DataFrame(
    [(p[0],p[1],count) for p, count in pair_counts.items()],
    columns = ['ProductA', 'ProductB', 'TimesBoughtTogether']
)
basket
```

```python
basket = basket[basket['TimesBoughtTogether'] > 1].sort_values(
    'TimesBoughtTogether', ascending=False

)
basket
```
