```python
import pandas as pd
```

```python
from google.colab import drive
drive.mount ('/content/drive')
df = pd.read_excel('/content/drive/My Drive/Colab Notebooks/Superstore.xlsx')
df.head()
```

```python
RegionalSales = df.groupby('Region').agg(
    TotalSales = ('Sales','sum'), TotalProfit = ('Profit','sum')).reset_index()

RegionalSales
```

```python
RegionalSales['ProfitMargin'] = (RegionalSales['TotalProfit']*100/RegionalSales['TotalSales']).round(2)
RegionalSales
```
