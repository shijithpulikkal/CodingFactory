```python
import pandas as pd
```

```python
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_excel('/content/drive/My Drive/Colab Notebooks/Superstore.xlsx')
df.head()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df.head()
```

```python
df = df.sort_values('Order_Date')
df.head()
```

```python
first_seen = df.drop_duplicates(subset = 'Customer_ID', keep = 'first').copy()
first_seen['Sales_Month'] = first_seen['Order_Date'].dt.to_period('M')
first_seen
```

```python
new_customers_per_month = (first_seen.groupby('Sales_Month')['Customer_ID'].count().reset_index(name='NewCustomers'))
new_customers_per_month
```

```python
new_customers_per_month['CumulativeDistinctCustomers'] = new_customers_per_month['NewCustomers'].cumsum()
new_customers_per_month
```
