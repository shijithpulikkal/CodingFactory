```python
import pandas as pd
```

```python
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_excel(r'/content/drive/My Drive/Colab Notebooks/Superstore.xlsx')
df.head()
```

```python
df['Order_Date']= pd.to_datetime(df['Order_Date'])
df['Sales_Month']= df['Order_Date'].dt.to_period('M')
df.head()
```

```python
Monthly_Sales = df.groupby('Sales_Month')['Sales'].sum().reset_index()
Monthly_Sales
```

```python
Monthly_Sales = Monthly_Sales.sort_values('Sales_Month').reset_index(drop=True)
Monthly_Sales
```

```python
Monthly_Sales['Moving_Average_3Months'] = (Monthly_Sales['Sales'].rolling(window=3,min_periods=1).mean().round(2))
Monthly_Sales
```
