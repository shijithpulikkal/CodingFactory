```python
import pandas as pd
```

```python
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_excel(r'/content/drive/My Drive/Colab Notebooks/Superstore.xlsx')
df.head()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['SalesMonth'] = df['Order_Date'].dt.to_period('M')
df['SalesMonth2'] = df['Order_Date'].dt.month
df.head()
```

```python
monthly_sales= df.groupby('SalesMonth')['Sales'].sum().reset_index().sort_values('SalesMonth')
monthly_sales
```

```python
monthly_sales['PrevMOnthSales'] = monthly_sales['Sales'].shift(1)
monthly_sales
```

```python
monthly_sales['Mom_Growth_pct'] = ((monthly_sales['Sales']-monthly_sales['PrevMOnthSales'])*100 / monthly_sales['PrevMOnthSales']).round(2)
monthly_sales
```
