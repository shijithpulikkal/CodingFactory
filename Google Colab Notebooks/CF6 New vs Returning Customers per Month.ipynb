```python
import pandas as pd
```

```python
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_excel(r'/content/drive/My Drive/Colab Notebooks/Superstore.xlsx')
df.head(5)
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
first_purchase = df.groupby('Customer_ID')['Order_Date'].min().reset_index()
first_purchase
```

```python
first_purchase.columns=['Customer_ID','First_Order_Date']
first_purchase
```

```python
merged = df.merge(first_purchase, on = 'Customer_ID')
merged
```

```python
merged['Order_Month'] = merged['Order_Date'].dt.to_period('M')
merged['First_Order_Month'] = merged['First_Order_Date'].dt.to_period('M')
merged
```

```python
merged['Customer_Type'] = merged.apply(
    lambda row: 'New' if row['Order_Month'] == row['First_Order_Month'] else 'Returning', axis = 1)
merged
```

```python
summary = (
    merged.groupby(['Order_Month','Customer_Type'])
    .agg(CustomerCount = ('Customer_ID','nunique'), TotalSales = ('Sales','sum'))).reset_index()

summary
```
