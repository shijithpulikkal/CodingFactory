```python
import pandas as pd
```

```python
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_excel('/content/drive/My Drive/Colab Notebooks/Superstore.xlsx')
df.head()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
```

```python
first_purchase = df.groupby('Customer_ID')['Order_Date'].min().dt.to_period('M')
first_purchase.name = 'CohortMonth'
first_purchase
```

```python
merged = df.merge(first_purchase, on ='Customer_ID')
merged['OrderMonth'] = merged['Order_Date'].dt.to_period('M')
merged.head()
```

```python
merged['MonthsSinceFirstPurchase'] = (
    (merged['OrderMonth']-merged['CohortMonth']).apply(lambda x: x.n)
    )
merged
```

```python
Cohort_summary = merged.groupby(['CohortMonth','MonthsSinceFirstPurchase']).agg(
    CohortRevenue=('Sales','sum'),
    ActiveCutomers=('Customer_ID','nunique')
).reset_index()

Cohort_summary
```

```python
cohort_pivot = Cohort_summary.pivot(
    index='CohortMonth', columns='MonthsSinceFirstPurchase', values='CohortRevenue'
)

cohort_pivot
```
