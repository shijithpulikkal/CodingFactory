```python
import pandas as pd
```

```python
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_excel(r'/content/drive/My Drive/Colab Notebooks/Superstore.xlsx')
df.head()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Sales_Year'] = df['Order_Date'].dt.year
df.head()
```

```python
yearlySales = df.groupby('Sales_Year')['Sales'].sum().reset_index()
yearlySales = yearlySales.sort_values(by='Sales_Year').reset_index(drop=True)
yearlySales

```

```python
yearlySales['PrevYearSales'] = yearlySales['Sales'].shift(1)
yearlySales
```

```python
yearlySales['YOY%'] = (((yearlySales['Sales'] - yearlySales['PrevYearSales'])/yearlySales['PrevYearSales'])*100).round(2)
yearlySales
```
