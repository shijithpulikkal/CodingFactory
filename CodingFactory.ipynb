```python
# @title
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_excel('/content/drive/My Drive/Colab Notebooks/Superstore.xlsx')
df.head()
```

```python
df['OrderDate'] = pd.to_datetime(df['Order Date'])
df['Sales_Year'] = df['OrderDate'].dt.year

df.head()
```

```python
yearly_spend = (df.groupby(['Sales_Year','Customer Name'])['Sales'].sum().reset_index(name='YearlySales'))
df.head()

idx = yearly_spend.groupby('Sales_Year')['YearlySales'].idxmax()
top_customers = yearly_spend.loc[idx].sort_values('Sales_Year').reset_index(drop=True)
top_customers
```
